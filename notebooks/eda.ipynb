```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# set theme for seaborn
sns.set_theme(style="whitegrid")
```

```python
FILE_PATH = "../data/processed/clean_ddos_traffic.csv"

df = pd.read_csv(FILE_PATH)
```

```python
df.columns = df.columns.str.strip()
```

```python
print(f"Data shape is {df.shape}")

display(df.head())

print("\n Data Info")
display(df.info())

print("\n Summary Statistics")
display(df.describe())

print("\n Missing Values (NaN)")
print(df.isna().sum()[df.isna().sum() > 0]) # show columns with missing data

# check infinity values 
numeric_cols = df.select_dtypes(include=np.number).columns
inf_count = np.isinf(df[numeric_cols]).sum().sum()
print(f"\n Infinities Count: {inf_count}")
```

```python
print("Target Variable ('Label') Distribution:")
print(df['Label'].value_counts())

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Label')
plt.title("Friday Afternoon Traffic: Normal vs. DDoS")
plt.show()
```
